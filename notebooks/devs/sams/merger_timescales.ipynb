```python
def print_tau_and_fem_from_a0(M,q,a0,return_only=False):
    m1,m2 = utils.m1m2_from_mtmr(M,q)
    _tauGW_yr = utils.time_to_merge_at_sep(m1, m2, a0) / YR # seconds
    _forb = utils.kepler_freq_from_sepa(M, a0)
    _fem = 2*_forb
    _fem_yrinv = _fem * YR
    if not return_only:
        print(f"{_tauGW_yr=:.4g} tauGW[Gyr]={_tauGW_yr/1e9:.2g} {_fem=:.2g} {_fem_yrinv=:.2g}")
    return _tauGW_yr
    
def print_tau_and_a0_from_fem(M,q,fem,return_only=False):
    m1,m2 = utils.m1m2_from_mtmr(M,q)
    _forb = fem / 2
    _a0 = utils.kepler_sepa_from_freq(M, _forb)
    _tauGW_yr = utils.time_to_merge_at_sep(m1, m2, _a0) / YR # seconds
    if not return_only:
        print(f"{_tauGW_yr=:.4g} tauGW[Gyr]={_tauGW_yr/1e9:.2g} {_a0/PC=:.2}")
    return _tauGW_yr
    
def eta_from_q(q):
    return q/(1+q)**2
```

```python
def Mmin_gw_mrg(z,Tobs,q):
    tauGW_yr_M9 = print_tau_and_a0_from_fem(1.0e9*MSOL,1.0,1/YR,return_only=True)
    tlook_yr = (cosmo.age(0).value - cosmo.age(z).value)*1e9
    eta = eta_from_q(q)
    Mmin_9 = (tauGW_yr_M9 / tlook_yr / (4*eta))**(3/5) * (Tobs*(1+z))**(8/5) 
    return Mmin_9 * 1e9
```
