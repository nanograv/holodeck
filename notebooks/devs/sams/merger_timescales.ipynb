```python
import numpy as np
from holodeck import cosmo, utils, plot
from holodeck.constants import MSOL, PC, YR, MPC, GYR, SPLC, SCHW, NWTG
from utils import _AGE_UNIVERSE_GYR
```

```python
def print_tau_and_fem_from_a0(M,q,a0):
    m1,m2 = utils.m1m2_from_mtmr(M,q)
    _tauGW_yr = utils.time_to_merge_at_sep(m1, m2, a0) / YR # seconds
    _forb = utils.kepler_freq_from_sepa(M, a0)
    _fem = 2*_forb
    _fem_yrinv = _fem * YR
    print(f"{_tauGW_yr=:.4g} tauGW[Gyr]={_tauGW_yr/1e9:.2g} {_fem=:.2g} {_fem_yrinv=:.2g}")
    return _tauGW_yr
    
def print_tau_and_a0_from_fem(M,q,fem):
    m1,m2 = utils.m1m2_from_mtmr(M,q)
    _forb = fem / 2
    _a0 = utils.kepler_sepa_from_freq(M, _forb)
    _tauGW_yr = utils.time_to_merge_at_sep(m1, m2, _a0) / YR # seconds
    print(f"{_tauGW_yr=:.4g} tauGW[Gyr]={_tauGW_yr/1e9:.2g} {_a0/PC=:.2}")
    return _tauGW_yr
    
def eta_from_q(q):
    return q/(1+q)**2
```

```python
M = 1.0e9*MSOL
q = 1.0
print_tau_and_a0_from_fem(M,q,1/(20*YR))
```
